# Welcome to Token Data

This package provides a unified interface to download data from different sources, starting with Coinbase and Hyperliquid.

## Installation

You can install the package using pip:

```bash
pip install token-data
```

## Usage

Here's how to use the `CoinbaseDataSource` and `HyperliquidDataSource` classes to fetch data.

````python
from token_data.coinbase import CoinbaseDataSource
from token_data.hyperliquid import HyperliquidDataSource
from pathlib import Path
from datetime import datetime, timedelta

# Create a data folder
data_folder = Path('data')
data_folder.mkdir(exist_ok=True)

# Initialize Coinbase data source
coinbase = CoinbaseDataSource(data_folder / 'coinbase')

# Get tokens
coinbase_tokens = coinbase.get_tokens()
print(f"First 5 Coinbase tokens: {coinbase_tokens[:5]}")

# Get spot prices
end_date = datetime.now()
start_date = end_date - timedelta(days=5)
btc_spot_prices = coinbase.get_spot_prices('BTC-USD', start_date.isoformat(), end_date.isoformat(), '1h')
print("BTC spot prices from Coinbase:")
print(btc_spot_prices.head())

# Initialize Hyperliquid data source
hyperliquid = HyperliquidDataSource(data_folder / 'hyperliquid')

# Get tokens
hyperliquid_tokens = hyperliquid.get_tokens()
print(f"First 5 Hyperliquid tokens: {hyperliquid_tokens[:5]}")

# Get perpetual prices
eth_perp_prices = hyperliquid.get_perp_prices('ETH', start_date.isoformat(), end_date.isoformat(), '1h')
print("ETH perpetual prices from Hyperliquid:")
print(eth_perp_prices.head())

# Get funding rates
eth_funding_rates = hyperliquid.get_funding_rates('ETH', start_date.isoformat(), end_date.isoformat())
print("ETH funding rates from Hyperliquid:")
print(eth_funding_rates.head())
````
